# 03 — Income distributions and CCDF
Restores the three representations present in the original work: ordinary CCDF, log-log CCDF, and the legacy double-log transformation. It also compares nominal and adjusted income.


```python
from pathlib import Path
from pnad_income.io import load_refined_panel
from pnad_income.distributions import build_ccdf_by_year, compare_income_measures_ccdf
from pnad_income.plotting import plot_ccdf_grid, plot_measure_comparison

panel = load_refined_panel(Path("../data/processed"))
```

```python
ccdf_nominal = build_ccdf_by_year(panel, value_col="income")
ccdf_adjusted = build_ccdf_by_year(panel, value_col="income_adj")
```

```python
fig_linear = plot_ccdf_grid(ccdf_nominal, transform="linear")
fig_loglog = plot_ccdf_grid(ccdf_nominal, transform="loglog")
fig_double_log = plot_ccdf_grid(ccdf_nominal, transform="double_log", min_ccdf=1)
```
